```python
def funct(n, df, ran):
    np.random.seed(ran)
    idx = np.arange(n)
    np.random.shuffle(idx)
    
    
    df_train = df.iloc[idx[:n_train]]
    df_val = df.iloc[idx[n_train:n_train+n_val]]
    df_test = df.iloc[idx[n_train+n_val:]]
    
    df_train_0 = df_train.fillna(0)
    df_val_0 = df_val.fillna(0)
    df_test_0 = df_test.fillna(0)

    y_train = np.array(df_train.fuel_efficiency_mpg.values)
    y_val = np.array(df_val.fuel_efficiency_mpg.values)
    y_test = np.array(df_test.fuel_efficiency_mpg.values)
    
    X_train_0 = df_train_0.drop(labels = 'fuel_efficiency_mpg', axis=1)
    X_val_0 = df_val_0.drop(labels = 'fuel_efficiency_mpg', axis=1)
    X_test_0 = df_test_0.drop(labels = 'fuel_efficiency_mpg', axis=1)

    w0, w = train_linear_regression(X_train_0, y_train)
    y_pred_0 = w0 + X_val_0.dot(w)

    return rmse(y_val, y_pred_0)
    
```

```python
n = len(df)
n_val = int(n*0.2)
n_test = int(n*0.2)
n_train = n-n_val-n_test

np.random.seed(42)
idx = np.arange(n)
np.random.shuffle(idx)


df_train = df.iloc[idx[:n_train]]
df_val = df.iloc[idx[n_train:n_train+n_val]]
df_test = df.iloc[idx[n_train+n_val:]]

df_train_0 = df_train.fillna(0)
df_val_0 = df_val.fillna(0)
df_test_0 = df_test.fillna(0)

mean = df_train['horsepower'].mean()

df_train_m = df_train.fillna(mean)
df_val_m = df_val.fillna(mean)
df_test_m = df_test.fillna(mean)
```

```python
y_train = np.array(df_train.fuel_efficiency_mpg.values)
y_val = np.array(df_val.fuel_efficiency_mpg.values)
y_test = np.array(df_test.fuel_efficiency_mpg.values)
```

```python
X_train_0 = df_train_0.drop(labels = 'fuel_efficiency_mpg', axis=1)
X_val_0 = df_val_0.drop(labels = 'fuel_efficiency_mpg', axis=1)
X_test_0 = df_test_0.drop(labels = 'fuel_efficiency_mpg', axis=1)

X_train_m = df_train_m.drop(labels = 'fuel_efficiency_mpg', axis=1)
X_val_m = df_val_m.drop(labels = 'fuel_efficiency_mpg', axis=1)
X_test_m = df_test_m.drop(labels = 'fuel_efficiency_mpg', axis=1)
```

```python
def train_linear_regression(X, y):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    return w_full[0], w_full[1:]
```

```python
w0, w = train_linear_regression(X_train_0, y_train)
y_pred_0 = w0 + X_val_0.dot(w)
```

```python
w0, w = train_linear_regression(X_train_m, y_train)
y_pred_m = w0 + X_val_m.dot(w)
```

```python
def rmse(y, y_pred):
    se = (y - y_pred) ** 2
    mse = se.mean()
    return np.sqrt(mse)
```

```python
round(rmse(y_val, y_pred_0),3)
```

```python
round(rmse(y_val, y_pred_m), 3)
```

```python
r = [0, 0.01, 0.1, 1, 5, 10, 100]
```

```python
def train_linear_regression_reg(X, y, r=0.001):
    ones = np.ones(X.shape[0])
    X = np.column_stack([ones, X])

    XTX = X.T.dot(X)
    XTX = XTX + r * np.eye(XTX.shape[0])

    XTX_inv = np.linalg.inv(XTX)
    w_full = XTX_inv.dot(X.T).dot(y)
    
    return w_full[0], w_full[1:]
```

```python
for reg in r:
    w0, w = train_linear_regression_reg(X_train_0, y_train, reg)
    y_pred = w0 + X_val_0.dot(w)
    print (reg, round(rmse(y_val, y_pred),4))
```
